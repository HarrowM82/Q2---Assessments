```python
def add(a, b):
  return a + b


def subtract(a, b):
  return a - b


def multiply(a, b):
  return a * b


def divide(a, b):
  return a / b if b != 0 else "Cannot divide by zero"


def main():
  a = float(input("First number: "))
  op = input("Operator (+, -, *, /): ")
  b = float(input("Second number: "))

  operations = {"+": add, "-": subtract, "*": multiply, "/": divide}

  if op in operations:
    func = operations[op]
    result = func(a, b)
    print("Result:", result)
  else:
    print("Invalid operator")


main()
```
