```python
def display(scores, label="Scores"):
    print(f"\n--- {label} ---")
    for i, s in enumerate(scores):
        print(f"[{i}]: {s}")

def search(scores):
    val = float(input("Score to search: "))
    if val in scores:
        print(f"Found {val} at index {scores.index(val)}")
    else:
        print("Score not found.")

def average(scores):
    print(f"Average Score: {sum(scores) / len(scores):.2f}")

def update(scores):
    idx = int(input(f"Enter index (0-{len(scores)-1}): "))
    if 0 <= idx < len(scores):
        scores[idx] = float(input("Enter new score: "))
        print("Score updated!")
    else:
        print("Invalid index.")

def main():
    scores = [85, 92, 78, 90, 88]

    while True:
        print("\n=== Student Score Analyzer ===")
        print("1. Display Scores\n2. Search Score\n3. Compute Average Score")
        print("4. Update Score\n5. Display Updated Scores\n6. Exit")
        choice = input("Select operation (1-6): ")

        if choice == "1": display(scores, "Current Scores")
        elif choice == "2": search(scores)
        elif choice == "3": average(scores)
        elif choice == "4": update(scores)
        elif choice == "5": display(scores, "Updated Scores")
        elif choice == "6": break
        else: print("Invalid choice.")

main()
```
